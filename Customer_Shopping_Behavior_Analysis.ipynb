```python
pd.read_sql(
    "SELECT AVG(purchase_amount) FROM customer",
    engine
)
```

```python
pd.read_sql("""
SELECT customer_id,
       purchase_amount
FROM customer
ORDER BY purchase_amount DESC
LIMIT 10
""", engine)
```

```python
pd.read_sql("""
SELECT gender,
       AVG(purchase_amount) AS avg_purchase
FROM customer
GROUP BY gender
""", engine)
```

```python
pd.read_sql("""
SELECT category,
       COUNT(*) AS total_customers
FROM customer
GROUP BY category
ORDER BY total_customers DESC
""", engine)
```

```python
pd.read_sql("""
SELECT *
FROM customer
WHERE location='Kentucky'
LIMIT 10
""", engine)
```
